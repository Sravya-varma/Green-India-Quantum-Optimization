```python

from qiskit_optimization import QuadraticProgram

# Create the optimization problem
problem = QuadraticProgram("Green_India_Optimization")

initiatives = [
    "Tree_Plantation",
    "Solar_Energy",
    "Waste_Management",
    "Pollution_Control",
    "Water_Conservation"
]

costs = [3, 4, 2, 3, 2]
impacts = [8, 9, 5, 7, 6]
budget = 7

# Binary decision variables:
# 1 = select initiative, 0 = do not select
for name in initiatives:
    problem.binary_var(name=name)

# Maximize the estimated impact score
problem.maximize(
    linear=dict(zip(initiatives, impacts))
)

# Keep total cost within the budget
problem.linear_constraint(
    linear=dict(zip(initiatives, costs)),
    sense="<=",
    rhs=budget,
    name="Budget_Constraint"
)

print(problem.prettyprint())
```

```python

from qiskit.primitives import StatevectorSampler
from qiskit_optimization.minimum_eigensolvers import QAOA
from qiskit_optimization.algorithms import MinimumEigenOptimizer
from qiskit_optimization.optimizers import COBYLA

# Create the QAOA algorithm
sampler = StatevectorSampler(seed=42)

qaoa = QAOA(
    sampler=sampler,
    optimizer=COBYLA(maxiter=100),
    reps=1
)

# Connect QAOA to the optimization problem
quantum_optimizer = MinimumEigenOptimizer(qaoa)

print("Starting Green India QAOA optimization...")

# Run the optimization
qaoa_result = quantum_optimizer.solve(problem)

print("\nQAOA RESULT")
print("-----------")
print(qaoa_result.prettyprint())
```

```python

import pandas as pd

# Extract the actual QAOA solution
solution_values = qaoa_result.x

initiative_names = [
    "Tree Plantation",
    "Solar Energy",
    "Waste Management",
    "Pollution Control",
    "Water Conservation"
]

costs = [3, 4, 2, 3, 2]
impacts = [8, 9, 5, 7, 6]
budget = 7

# Convert solution values into binary decisions
selected_values = [
    int(round(value)) for value in solution_values
]

# Calculate actual result metrics
quantum_cost = sum(
    selected_values[i] * costs[i]
    for i in range(len(initiative_names))
)

quantum_score = sum(
    selected_values[i] * impacts[i]
    for i in range(len(initiative_names))
)

print("GREEN INDIA - QAOA RESULT")
print("-------------------------")
print("Selected cost:", quantum_cost)
print("Estimated impact score:", quantum_score)
print("Budget:", budget)
print("Budget feasible:", quantum_cost <= budget)

print("\nSelected initiatives:")
for i, selected in enumerate(selected_values):
    if selected:
        print("-", initiative_names[i])

# Create a detailed table of the actual QAOA result
quantum_details = pd.DataFrame({
    "Initiative": initiative_names,
    "Cost": costs,
    "Estimated Impact Score": impacts,
    "Selected": selected_values
})

display(quantum_details)
```

```python

import pandas as pd
from itertools import product

# Green India project data (illustrative values)
initiatives = [
    "Tree Plantation",
    "Solar Energy",
    "Waste Management",
    "Pollution Control",
    "Water Conservation"
]

costs = [3, 4, 2, 3, 2]
impacts = [8, 9, 5, 7, 6]
budget = 7

# Find the best classical solution
best_score = -1
best_selection = []

for selection in product([0, 1], repeat=len(initiatives)):
    total_cost = sum(
        selection[i] * costs[i] for i in range(len(initiatives))
    )
    total_score = sum(
        selection[i] * impacts[i] for i in range(len(initiatives))
    )

    if total_cost <= budget and total_score > best_score:
        best_score = total_score
        best_selection = list(selection)

classical_cost = sum(
    best_selection[i] * costs[i] for i in range(len(initiatives))
)

# Read the QAOA result from your previous cell
quantum_selection = [int(round(x)) for x in qaoa_result.x]

quantum_cost = sum(
    quantum_selection[i] * costs[i] for i in range(len(initiatives))
)
quantum_score = sum(
    quantum_selection[i] * impacts[i] for i in range(len(initiatives))
)

# Display comparison
comparison = pd.DataFrame({
    "Method": ["Classical Optimization", "Quantum QAOA"],
    "Total Cost": [classical_cost, quantum_cost],
    "Estimated Impact Score": [best_score, quantum_score],
    "Within Budget": [
        classical_cost <= budget,
        quantum_cost <= budget
    ]
})

print("GREEN INDIA: CLASSICAL VS QUANTUM")
print("Budget:", budget)
display(comparison)

print("\nClassical selected initiatives:")
for i, selected in enumerate(best_selection):
    if selected:
        print("-", initiatives[i])

print("\nQuantum QAOA selected initiatives:")
for i, selected in enumerate(quantum_selection):
    if selected:
        print("-", initiatives[i])
```
